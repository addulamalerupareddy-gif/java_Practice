```python
import numpy as np

arr8 = np.array([10, 20, 30, 40, 50], dtype=np.int8)
arr16 = np.array([10, 20, 30, 40, 50], dtype=np.int16)
arr32 = np.array([10, 20, 30, 40, 50], dtype=np.int32)
arr64 = np.array([10, 20, 30, 40, 50], dtype=np.int64)

for arr in [arr8, arr16, arr32, arr64]:
    print("dtype:", arr.dtype)
    print("itemsize:", arr.itemsize, "bytes")
    print("nbytes:", arr.nbytes, "bytes")
    print()
```

```python
import numpy as np

arr16 = np.array([10.5, 20.5, 30.5], dtype=np.float16)
arr32 = np.array([10.5, 20.5, 30.5], dtype=np.float32)
arr64 = np.array([10.5, 20.5, 30.5], dtype=np.float64)

obj_arr = np.array(["Python", 100, 3.14, True], dtype=object)

for arr in [arr16, arr32, arr64, obj_arr]:
    print("Array:", arr)
    print("dtype:", arr.dtype)
    print("itemsize:", arr.itemsize)
    print("nbytes:", arr.nbytes)
    print()
```
