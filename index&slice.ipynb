to combine multiple condition

-use only operator symbol '&' for and i for or

-always enclose conditions in parenthesis(condition)

Get the numbers divisible by  either  3 and 5


```python
arr2[(arr2%3==0)|(arr2%5==0)]
```

Get the numbers divisible by both 3 and 5

```python
arr2[(arr2%3==0)&(arr2%5==0)]
```

get the numbers between 1,100.  neither divisible by 3 nor divisible by 5

```python
arr2[~((arr2%3==0)|(arr2%5==0))]
```
