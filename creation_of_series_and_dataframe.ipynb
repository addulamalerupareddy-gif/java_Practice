in a dataframe every column may assume different datatype,based on the variable it assumes

-to check 'dtype and assign 'dtype use '.dtypes'

-.dtypes always returns a series with values

```python
df5.dtypes
```

```python
#modify the column,row
df5=pd.DataFrame(np.arange(1,26).reshape(5,5),
                 index=['r1','r2','r3','r4','r5'],
                 columns=['c1','c2','c3','c4','c5'],
                 dtype='uint8')
df5
```

```python
#change the datatype of every columns in dataframe
df5.astype({'c1':'uint8','c2':'uint16','c3':'uint32','c4':'uint64','c5':'int8'})
df5
```

```python
df5.dtypes
```
