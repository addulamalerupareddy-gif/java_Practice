```python
(26)
def factorial(n):
  if n==0 or n==1:
    return 1
  return n*factorial(n-1)
print(factorial(5))
```

```python
(27)
def fib(n):
  if n==0:
    return 0
  if n==1:
    return 1
  return fib(n-1)+fib(n-2)
print(fib(6))
```

```python
(28)
def sum_natural(n):
  if n==0:
    return 0
  return n+sum_natural(n-1)
print(sum_natural(5))
```

```python
(29)
def power(a,n):
  if n==0:
    return 1
  return a*power(a,n-1)
print(power(2,5))
```
