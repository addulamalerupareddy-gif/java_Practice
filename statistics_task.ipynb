# Descriptive Statistics: Understanding Data Before Machine Learning

Descriptive statistics helps us **summarize, understand, and describe data** before building a Machine Learning model.

For example, if a shop records daily sales as:

**₹10, ₹12, ₹12, ₹15, ₹21**

we can use descriptive statistics to answer:

* What is the typical sales amount? → **Mean, Median, Mode**
* How much do sales vary? → **Range, Variance, Standard Deviation**
* Where are the middle 50% of values? → **Quartiles, IQR**
* Is the data balanced or skewed? → **Skewness**
* Does it contain unusually heavy tails/extreme values? → **Kurtosis**

---

# 1️⃣ Measures of Central Tendency

## What does "finding the center" mean?

The **center of data** is a value that gives us an idea of what is typical or representative of the dataset.

The three main measures are:

| Measure    | Simple meaning                  |
| ---------- | ------------------------------- |
| **Mean**   | Average                         |
| **Median** | Middle value                    |
| **Mode**   | Most frequently occurring value |

---

## 1. Mean

### Simple explanation

The **mean** is the average of all values.

genui{"learning_viz":{"type_id":"ARITHMETIC_MEAN"}}

### Why is Mean important?

It gives us one number representing the overall level of the data.

It is useful for:

* Average sales
* Average marks
* Average temperature
* Average customer spending

### Formula

$$
Mean=\frac{\sum x}{n}
$$

Where:

* \(x\) = each value
* \(n\) = number of values

### Small problem

Suppose daily sales are:

**10, 12, 12, 15, 21**

### Step 1: Add all values

$$
10+12+12+15+21=70
$$

### Step 2: Count the values

$$
n=5
$$

### Step 3: Divide

$$
Mean=\frac{70}{5}=14
$$

### Answer

**Mean = 14**

### Interpretation

The average daily sales value is **₹14**.

### Business example

A company calculates average monthly sales to understand its overall sales performance.

### Limitation

Mean is sensitive to **extreme values**.

For example:

**10, 12, 12, 15, 100**

The 100 can increase the mean considerably.

---

# 2. Median

### Simple explanation

The **median** is the middle value after arranging the data in ascending order.

### Why is Median important?

It is especially useful when the dataset contains **outliers or is skewed**.

### Problem

Data:

**10, 12, 12, 15, 21**

There are 5 values.

The middle value is:

**12**

Therefore:

$$
Median=12
$$

### Interpretation

The median tells us the central position of the ordered data.

### If there are an even number of values

Example:

**10, 12, 15, 20**

The two middle values are 12 and 15.

$$
Median=\frac{12+15}{2}=13.5
$$

### Business example

Suppose house prices are:

**₹20 lakh, ₹25 lakh, ₹30 lakh, ₹35 lakh, ₹5 crore**

The ₹5 crore house is an extreme value.

For understanding the **typical house price**, a business may prefer the median because it is less affected by that extreme value.

### Practical use

Median is commonly useful for:

* Salaries
* Income
* House prices
* Customer spending
* Delivery times

when the data is skewed.

---

# 3. Mode

### Simple explanation

The **mode** is the value that occurs most often.

### Problem

Data:

**10, 12, 12, 15, 21**

Count the values:

* 10 → 1 time
* 12 → **2 times**
* 15 → 1 time
* 21 → 1 time

Therefore:

$$
Mode=12
$$

### Interpretation

12 is the most frequently occurring value.

### Business example

A clothing store sells:

* Small → 40
* Medium → **100**
* Large → 60

The **mode is Medium**.

The store knows that Medium is the most popular size.

### Practical use

Mode is useful for categorical information such as:

* Most popular product
* Most common size
* Most common payment method
* Most common city

---

# Mean vs Median vs Mode

For:

**10, 12, 12, 15, 21**

| Measure | Result | Meaning     |
| ------- | -----: | ----------- |
| Mean    |     14 | Average     |
| Median  |     12 | Middle      |
| Mode    |     12 | Most common |

### Easy memory trick

> **Mean = Average**
> **Median = Middle**
> **Mode = Most common**

---

# Why Median is preferred for skewed data

Consider salaries:

**₹20,000, ₹22,000, ₹25,000, ₹28,000, ₹2,00,000**

The ₹2,00,000 salary is very high compared with the others.

The **mean gets pulled toward ₹2,00,000**.

The median stays closer to the salary of a typical employee.

Therefore:

> **Businesses often prefer Median when data is highly skewed or contains extreme values.**

# 2️⃣ Measures of Dispersion

Central tendency tells us **where the center is**.

But knowing only the average is **not enough**.

Consider:

### Company A

**48, 49, 50, 51, 52**

### Company B

**10, 30, 50, 70, 90**

Both have:

$$
Mean=50
$$

But Company A is much more consistent.

Company B has much greater variation.

So we need **dispersion measures**.

---

# 4. Range

### Simple explanation

Range tells us how far the data extends from the smallest to the largest value.

### Formula

$$
Range=Maximum-Minimum
$$

### Problem

Data:

**10, 12, 12, 15, 21**

Maximum:

$$
21
$$

Minimum:

$$
10
$$

Therefore:

$$
Range=21-10=11
$$

### Answer

**Range = 11**

### Interpretation

The difference between the highest and lowest values is **11**.

### Business example

Delivery times:

**2, 3, 4, 5, 10 days**

$$
Range=10-2=8
$$

The delivery time varies across an 8-day range.

### Limitation

Range only considers the **minimum and maximum**. It ignores the values between them.

---

# 5. Variance

### Simple explanation

Variance measures how much the values **spread around the mean**.

A high variance means greater variability.

genui{"learning_viz":{"type_id":"VARIANCE"}}

### Formula

For population data:

$$
\sigma^2=\frac{\sum(x-\mu)^2}{N}
$$

### Problem

Use:

**10, 12, 12, 15, 21**

Mean:

$$
\mu=14
$$

### Step 1: Find deviations

| Value | Value − Mean |
| ----: | -----------: |
|    10 |           -4 |
|    12 |           -2 |
|    12 |           -2 |
|    15 |            1 |
|    21 |            7 |

### Step 2: Square them

| Value | Deviation | Squared deviation |
| ----: | --------: | ----------------: |
|    10 |        -4 |                16 |
|    12 |        -2 |                 4 |
|    12 |        -2 |                 4 |
|    15 |         1 |                 1 |
|    21 |         7 |                49 |

### Step 3: Add

$$
16+4+4+1+49=74
$$

### Step 4: Divide by 5

$$
Variance=\frac{74}{5}=14.8
$$

### Answer

**Population Variance = 14.8**

### Interpretation

The data has measurable variability around its mean of 14.

### Business example

A company can use variance to understand how much daily sales fluctuate.

### Limitation

Variance uses **squared units**, so it can be harder to interpret directly.

> For sample data, the usual variance formula divides by \(n-1\) instead of \(n\).

---

# 6. Standard Deviation

### Simple explanation

Standard deviation tells us how spread out the data is around the mean, in the **same units as the original data**.

genui{"learning_viz":{"type_id":"STANDARD_DEVIATION"}}

### Formula

$$
SD=\sqrt{Variance}
$$

From above:

$$
Variance=14.8
$$

Therefore:

$$
SD=\sqrt{14.8}
$$

$$
SD\approx3.85
$$

### Answer

**Standard Deviation ≈ 3.85**

### Interpretation

The values have a typical spread of roughly **3.85 units around the mean**.

### Business example

Two stores have the same average sales:

* Store A → SD = ₹500
* Store B → SD = ₹5,000

Store A has more consistent sales.

### Easy rule

> **Small SD → More consistency**
> **Large SD → More variability**

### Practical use

Standard deviation is useful in:

* Risk analysis
* Financial analysis
* Quality control
* Sales analysis
* Machine Learning

---

# 7. Quartiles

### Simple explanation

Quartiles divide ordered data into approximately **four parts**.

* **Q1** → 25th percentile
* **Q2** → 50th percentile / Median
* **Q3** → 75th percentile

### Problem

Data:

**10, 12, 12, 15, 21**

### Q2

The middle value:

$$
Q2=12
$$

### Q1

Lower half:

**10, 12**

$$
Q1=\frac{10+12}{2}=11
$$

### Q3

Upper half:

**15, 21**

$$
Q3=\frac{15+21}{2}=18
$$

### Answer

$$
Q1=11
$$

$$
Q2=12
$$

$$
Q3=18
$$

### Interpretation

* Q1 represents the lower quarter of the data.
* Q2 is the median.
* Q3 represents the upper quarter.

> Quartile calculation methods can differ slightly between statistical tools, so always check which method your software uses.

---

# 8. Interquartile Range — IQR

### Simple explanation

IQR tells us the spread of the **middle 50% of the data**.

### Formula

$$
IQR=Q3-Q1
$$

We have:

$$
Q3=18
$$

$$
Q1=11
$$

Therefore:

$$
IQR=18-11=7
$$

### Answer

**IQR = 7**

### Interpretation

The middle 50% of observations are spread across **7 units**.

---

# IQR for Outlier Detection

A common rule is:

### Lower boundary

$$
Q1-1.5(IQR)
$$

### Upper boundary

$$
Q3+1.5(IQR)
$$

For our example:

$$
Q1=11,\quad Q3=18,\quad IQR=7
$$

Lower:

$$
11-1.5(7)=11-10.5=0.5
$$

Upper:

$$
18+1.5(7)=18+10.5=28.5
$$

So values below **0.5** or above **28.5** would be considered potential outliers using this rule.

### Business example

Customer purchases:

**₹1,000, ₹1,200, ₹1,500, ₹1,700, ₹2,000, ₹25,000**

The ₹25,000 purchase may be an unusual transaction.

IQR can help analysts identify such observations.

### Practical use

IQR is particularly useful for **skewed data** because it is less affected by extreme values.

---

# Dispersion Summary

| Measure                | Meaning                              |
| ---------------------- | ------------------------------------ |
| **Range**              | Maximum − Minimum                    |
| **Variance**           | Squared spread around mean           |
| **Standard Deviation** | Spread around mean in original units |
| **Q1**                 | 25% point                            |
| **Q2**                 | 50% point / Median                   |
| **Q3**                 | 75% point                            |
| **IQR**                | Middle 50% spread                    |

# 3️⃣ Measures of Shape

Measures of shape tell us about the **distribution of the data**.

The two important concepts are:

1. **Skewness**
2. **Kurtosis**

---

# 9. Skewness

### Simple explanation

Skewness tells us whether the distribution is **symmetric or has a longer tail on one side**.

There are three main types:

* Symmetric
* Right-skewed
* Left-skewed

---

## A. Symmetric Distribution

Example:

**2, 3, 4, 5, 6**

```text
        *
      *   *
    *       *
  *           *
-----------------
  2  3  4  5  6
```

The data is approximately balanced around the center.

Here:

$$
Mean=4
$$

$$
Median=4
$$

### Interpretation

There is no strong tail toward either side.

### Business example

Measurements from a well-controlled manufacturing process may be approximately symmetric around the target.

---

# B. Right-Skewed Distribution

Example:

**2, 3, 3, 4, 20**

```text
****
******
********
********
****************------------>
2  3  4                  20
```

The tail extends toward the **right**.

Generally:

$$
Mean > Median
$$

### Beginner skewness calculation

One simple Pearson measure is:

$$
Skewness\approx\frac{3(Mean-Median)}{SD}
$$

Mean:

$$
\frac{2+3+3+4+20}{5}=6.4
$$

Median:

$$
3
$$

SD ≈ 6.83

Therefore:

$$
Skewness\approx
\frac{3(6.4-3)}{6.83}
$$

$$
\approx1.49
$$

### Interpretation

The positive value indicates **right skewness**.

### Business example

Income data can be right-skewed because a small number of people have very high incomes.

---

# C. Left-Skewed Distribution

Example:

**1, 10, 10, 11, 12**

```text
<--------------------
*
*
*
      ***************
1   10  10  11  12
```

The tail extends toward the **left**.

Generally:

$$
Mean < Median
$$

Mean:

$$
\frac{1+10+10+11+12}{5}=8.8
$$

Median:

$$
10
$$

SD ≈ 3.97

Therefore:

$$
Skewness\approx
\frac{3(8.8-10)}{3.97}
$$

$$
\approx-0.91
$$

### Interpretation

The negative value indicates **left skewness**.

### Business example

If most students score very high on an easy examination but a few score very low, the distribution may be left-skewed.

---

# Skewness Summary

| Skewness | Distribution |
| -------- | ------------ |
| ≈ 0      | Symmetric    |
| Positive | Right-skewed |
| Negative | Left-skewed  |

### Easy memory

> **Right tail → Positive skew**
> **Left tail → Negative skew**

---

# 10. Kurtosis

### Simple explanation

Kurtosis describes the **tail heaviness and tendency toward extreme observations** in a distribution.

It is useful for understanding whether data has unusually many extreme values.

---

## A. Normal Distribution

A normal distribution has a familiar bell shape:

```text
             *
           *   *
         *       *
       *           *
     *               *
---*-------------------*---
```

For a normal distribution:

### Ordinary kurtosis

$$
K=3
$$

### Excess kurtosis

$$
K_{excess}=0
$$

The excess-kurtosis scale is commonly used in data analysis.

### Interpretation

It represents the standard reference level of tail behavior.

---

# B. Heavy-Tailed Distribution

A heavy-tailed distribution has **more extreme observations** than a normal distribution.

Conceptually:

```text
*                         *
 **                     **
   ***                 ***
      *****       *****
          *******
```

Under the excess-kurtosis convention:

$$
K_{excess}>0
$$

### Interpretation

There is greater tendency for extreme values.

### Business example

Financial returns can sometimes have heavy tails because unusually large gains or losses occur more often than a normal model would suggest.

### Practical use

Heavy-tail information can be useful in:

* Financial risk analysis
* Fraud detection
* Anomaly detection
* Quality control

---

# C. Light-Tailed Distribution

A light-tailed distribution has **fewer extreme observations** than the normal reference.

Under the excess-kurtosis convention:

$$
K_{excess}<0
$$

### Simple example

Consider:

**1, 2, 3, 4, 5**

Mean:

$$
Mean=3
$$

Deviations:

**-2, -1, 0, 1, 2**

Fourth powers:

**16, 1, 0, 1, 16**

Fourth central moment:

$$
\mu_4=\frac{16+1+0+1+16}{5}
$$

$$
=\frac{34}{5}=6.8
$$

Variance:

$$
\sigma^2=2
$$

Therefore:

$$
\sigma^4=2^2=4
$$

Ordinary kurtosis:

$$
K=\frac{6.8}{4}=1.7
$$

Excess kurtosis:

$$
1.7-3=-1.3
$$

### Interpretation

The negative excess kurtosis indicates a **light-tailed** distribution under this calculation.

> Statistical software can use different sample kurtosis estimators, so exact values may differ.

---

# Kurtosis Summary

| Distribution     | Excess Kurtosis | Meaning                |
| ---------------- | --------------: | ---------------------- |
| Normal reference |             ≈ 0 | Standard tail behavior |
| Heavy-tailed     |             > 0 | More extreme values    |
| Light-tailed     |             < 0 | Fewer extreme values   |

### Important point

Kurtosis should **not be interpreted simply as "how tall the peak is."** In practical data analysis, its most useful intuition is about **tail behavior and extreme observations**.

# Why Descriptive Statistics Matters Before Machine Learning

Before building a Machine Learning model, we need to understand the dataset.

For example:

**Mean/Median** → What is a typical customer spending?

**Standard deviation** → How consistent is customer spending?

**IQR** → Are there unusual transactions?

**Skewness** → Is spending heavily concentrated toward one side?

**Kurtosis** → Are extreme transactions unusually common?

This understanding helps a Data Analyst/Data Scientist decide whether the data needs **cleaning, outlier treatment, transformation, scaling, or other preprocessing** before Machine Learning.
